## Question 3

```r
ice_cream <- read.csv("/anvil/projects/tdm/data/icecream/combined/products.csv")
summary(ice_cream$rating)
```

View summary of the rating column to get quartiles. This gives Min., 1st Qu., Median, Mean, 3rd Qu., Max. We will use these for our breaks in cut()

```r
breaks <- c(min(ice_cream$rating, na.rm = TRUE),
            quantile(ice_cream$rating, 0.25, na.rm = TRUE),
            median(ice_cream$rating, na.rm = TRUE),
            quantile(ice_cream$rating, 0.75, na.rm = TRUE),
            max(ice_cream$rating, na.rm = TRUE))
```

Define the breaks using Min, 1st Qu., Median, 3rd Qu., Max

```r
labels <- c("Wouldn't Recommend",
            "Needs Improvement",
            "Solid Choice",
            "Fan Favorite")
```

Define labels corresponding to each range

```r
ice_cream$rating_phrases <- cut(ice_cream$rating,
                                breaks = breaks,
                                labels = labels,
                                include.lowest = TRUE, right = TRUE)
```

```r
head(ice_cream$rating_phrases)
table(ice_cream$rating_phrases)
```

Below are "cut()" examples:

1. Create price categories for ice cream based on price column `ice_cream$price_category <- cut(ice_cream$price, breaks=5)`

2. Group ice creams by number of ingredients `ice_cream$ingredient_count_group <- cut(ice_cream$num_ingredients, breaks=c(0,5,10,15,20))`

3. Convert continuous ratings into simpler "high/medium/low" for plots `ice_cream$rating_simple <- cut(ice_cream$rating, breaks=3, labels=c("Low","Medium","High"))`

```r
my_flights <- read.csv("/anvil/projects/tdm/data/flights/subset/1997.csv")
table(my_flights$DepTime)
```

```r
breaks <- c(0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400)
labels <- c("Night", "Early Morning", "Morning", "Late Morning",
            "Afternoon", "Late Afternoon", "Evening", "Late Evening")
my_flights$depart_times <- cut(my_flights$DepTime,
                               breaks = breaks,
                               labels = labels,
                               include.lowest = TRUE, right = FALSE)
```

Break DepTime into ranges using "cut()"

```r
depart_table <- table(my_flights$depart_times)
depart_table
```

Table of departure counts

```r
barplot(depart_table,
        main = "Number of Flights by Departure Time (1997)",
        xlab = "Time of Day",
        ylab = "Number of Flights",
        col = "skyblue",
        las = 2)
```

Barplot for departure times

```r
names(depart_table)[which.max(depart_table)]
```

Time of day with most departure flights

```r
my_flights$arrival_times <- cut(my_flights$ArrTime,
                                breaks = breaks,
                                labels = labels,
                                include.lowest = TRUE, right = FALSE)
arrival_table <- table(my_flights$arrival_times)
arrival_table
```

```r
barplot(arrival_table,
        main = "Number of Flights by Arrival Time (1997)",
        xlab = "Time of Day",
        ylab = "Number of Flights",
        col = "lightgreen",
        las = 2)
```

Repeat same process for arrival times

```r
names(arrival_table)[which.min(arrival_table)]
```

Time of day with least arrival flights

Most flights depart in the "Morning" period, and fewest flights arrive during "Night". This indicates peak departure times are in the morning and early day, while late night is the dull period for arrivals.

```r
march <- subset(my_flights, Month == 3)
table(march$DayOfWeek)
```

```r
total_airtime_day <- tapply(my_flights$AirTime, my_flights$DayOfWeek, sum, na.rm = TRUE)
total_airtime_day
```

```r
flight_times <- tapply(my_flights$AirTime,
                       list(my_flights$DayOfWeek, my_flights$Month),
                       sum, na.rm = TRUE)
flight_times
```

```r
heatmap(flight_times, Rowv = NA, Colv = NA,
        main = "Total AirTime by Day of Week and Month (1997)",
        xlab = "Month", ylab = "Day of Week", col = heat.colors(12))
```

Heatmap of total AirTime by DayOfWeek (rows) and Month (columns)

```r
clean_flights <- na.omit(my_flights[, c("AirTime", "DayofMonth", "Month")])
```

```r
flight_dom <- tapply(clean_flights$AirTime,
                     list(clean_flights$DayofMonth, clean_flights$Month),
                     sum)
```

```r
heatmap(flight_dom,
        Rowv = NA, Colv = NA,
        scale = "none",
        main = "AirTime by DayofMonth and Month")
```

Some rows have NA values in AirTime, Month, or DayofMonth. We use na.omit() to drop those incomplete rows so that all vectors passed to `tapply()` have the same length and no missing values. This prevents the "arguments must have same length" error.

5.1: I preferred the day of week heatmap because it is easier to interpret and shows clear recurring weekly patterns. The day of month heatmap is more cluttered since months have different lengths (28–31 days), and flight counts vary irregularly, which makes it harder to compare across months. By contrast, the day of week heatmap immediately shows whether certain weekdays (e.g., Fridays or Sundays) tend to have more total air time consistently throughout the year.

5.2: Summer months (June, July, August) generally had higher total air time, likely because of increased travel during vacation season.

Around December, air time also spiked, probably due to holiday travel.

Some weekdays (like Fridays and Sundays) consistently had higher air times compared to mid-week days (like Tuesdays and Wednesdays), reflecting weekend and return-trip travel patterns.

Shorter months (like February) showed lower totals, but the weekly pattern still persisted.

5.3: I set Rowv = NA and Colv = NA in heatmap() to prevent clustering and to keep rows/columns in natural order (days and months).

Applied a color gradient (e.g., light to dark or blue to red) so that higher air times are visually distinct.

Added axis labels to clearly show which dimension is “Month” vs “Day of Week/Day of Month.”

Normalized or scaled values where needed to avoid extreme outliers overshadowing subtle differences.

Increased the font size for labels so patterns across days/months are readable without zooming.
