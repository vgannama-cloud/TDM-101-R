## If/else logic: mapping lightsaber color to Jedi role

**Differences between if/else and ifelse:**
1. Scope:
- if/else handles a single value or condition at a time.
- ifelse is vectorized and can handle an entire vector at once.

2. Output:
- if/else executes code blocks but doesn't directly return a vector.
- ifelse returns a vector with a result for each input element.

3. Usage:
- Use if/else for single decisions or inside loops.
- Use ifelse for quick classification or transformation of vectors without loops.

screen_time loop

```r
screen_time <- 10 
while(screen_time > 2) {
  print(paste("Screen time:", screen_time, "hours"))
  screen_time <- screen_time - 1
}
```

Differences between print(paste()) and cat():

print(paste()):
- Outputs a character vector
- Adds quotes and always prints a new line after each call
- Good for debugging

cat():
- Concatenates text without quotes
- You need to manually add newline (\n) if you want a line break
- Cleaner output for user-facing messages

```r
print(paste("Screen time:", 5, "hours"))
cat("Screen time:", 5, "hours\n")
```

```r
screen_time <- 10
sleep_time <- 2

while(screen_time > 2) {
  print(paste("Screen time:", screen_time, "hours,", "Sleep time:", sleep_time, "hours"))
  
  screen_time <- screen_time - 0.5
  sleep_time <- sleep_time + 0.5
}
```

## Looping through Star Wars characters to count non-human species

```r
characters <- read.csv("/anvil/projects/tdm/data/starwars/characters.csv")
head(characters)
```

```r
i <- 1
char_count <- 0
non_human_count <- 0 

while(char_count < 20) {
  
  if (characters$species[i] == "Human") {
    cat(char_count + 1, "This is a human\n") 
  } else {
    print("This is not a human")
    non_human_count <- non_human_count + 1
  }
  
  i <- i + 1
  char_count <- char_count + 1
}

cat("Number of non-human characters in first 20 is", non_human_count, "\n")
```

## Using a repeat loop to sum mental health scores

While Loop:
- Condition is checked before each iteration.
- Stops automatically when the condition becomes false.
- best used when the number of iterations is known or depends on a condition.

Repeat Loop:
- No initial condition. The loop runs atleast once.
- Must explicitly use a `break` statement to exit the loop.
- Useful when you want at least one iteration or need manual control over exiting the loop.

```r
i <- 1
total_count <- 0

repeat {
  student_score <- myDF$Mental_Health_Score[i]
  
  cat("Mental health of student", i, "is", student_score, "\n")
  cat("Current mental health score is", total_count, "\n\n")
    
  total_count <- total_count + student_score
  
  if (total_count >= 100) {
    print("ALLL DONEEEEE")
    break
  }
  i <- i + 1
}
```
