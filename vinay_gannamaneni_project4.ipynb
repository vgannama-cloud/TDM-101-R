## Question 4

```r
old_ages <- c(61, 62, 63, 64, 65, 66, 70, 70.5, 71, 74, 80)
```

Created vector for old ages exclusively

```r
old <- myDF$Age[!is.na(myDF$Age) & myDF$Age %in% old_ages]
```

To find which ages in the data set match these values

```r
my_vec1 <- myDF$Age[!is.na(myDF$Age) & myDF$Age <= 10]
my_vec2 <- myDF$Age[!is.na(myDF$Age) & myDF$Age > 10 & myDF$Age <= 60]
```

Created "my_vec1"(ages <= 10) & "my_vec2"(ages > 10 & <= 60)

```r
people <- c(my_vec1, my_vec2)
```

Combine my_vec1 and my_vec2 into "people"

```r
children <- people[people <= 20]
young    <- people[people > 20 & people <= 40]
adult    <- people[people > 40 & people <= 60]
```

Created children, young, adult categories from people

```r
table(children)
table(young)
table(adult)
table(old)
```

Breaking ages into ranges helps us see patterns more clearly. Instead of looking at every single age, we can group passengers into broader categories like kids, young adults, middle-aged adults, and seniors. This makes it much easier to compare survival rates and spot trends, since we’re focusing on meaningful life stages rather than scattered individual ages. It also makes visualizations and analysis more straightforward.

I’d prefer ranges like 0–12 (children), 13–18 (teenagers), 19–35 (young adults), 36–55 (adults), and 56+ (seniors). These feel closer to real-life stages of life and how people were probably viewed during the Titanic era. For example, children and teens might have been treated differently during evacuation, while seniors may have struggled more physically. Grouping this way makes the data not only easier to analyze but also more meaningful in terms of understanding the human side of the story.
